```python
Q1=df["Weekly_Study_Hours"].quantile(0.25)
Q3=df["Weekly_Study_Hours"].quantile(0.75)
IQR = Q3-Q1
Lower_Bound = Q1-1.5*IQR
Upper_Bound = Q3+1.5*IQR
Outlier=df[(df["Weekly_Study_Hours"] < Lower_Bound) | (df["Weekly_Study_Hours"] > Upper_Bound)]
print(f"Q1:{Q1}")
print(f"Q3:{Q3}")
print(f"Interquartile Range of Weekly_Study_Hours\n{IQR}")
print(f"Lower Bound:{Lower_Bound}")
print(f"Upper Bound:{Upper_Bound}")
print(f"No. of Outliers:{len(Outlier)}")
```
